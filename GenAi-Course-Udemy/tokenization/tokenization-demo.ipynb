```python
corpus = """Hello welcome to tokenization demo.
We will be performing tokenization on this corpus."""
```

```python
print(corpus)
```

```python
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')  # For newer versions of NLTK

# Tokenization
# Sentence --> Paragraph
from nltk.tokenize import sent_tokenize

sent_tokenize(corpus)
```

```python
# Paragraph --> Words
# Sentence --> Words

from nltk.tokenize import word_tokenize

word_tokenize(corpus)
```
