```python
import os
from dotenv import load_dotenv

load_dotenv(override=True)

os.environ['LANGSMITH_API_KEY'] = os.getenv("LANGSMITH_API_KEY")
os.environ['LANGSMITH_TRACING'] = "true"
os.environ['GEMINI_API_KEY'] = os.getenv("GEMINI_API_KEY")
os.environ['GEMINI_MODEL'] = os.getenv("GEMINI_MODEL")
```

# Create data point

```python
from langsmith import Client

client = Client()

# Define the dataset - these are your test data
dataset_name = "Simple Chatbot Evaluation"
dataset = client.create_dataset(dataset_name)

client.create_examples(
    dataset_id = dataset.id,
    examples=[
        {
            "input": {"question": "What is langchain?"},
            "output": {"answer": "A framework for building LLM applications"},
        },
        {
            "input": {"question": "What is langsmith?"},
            "output": {"answer": "A platform for observing and evaluating LLM application"},
        },
        {
            "input": {"question": "What is OpenAI?"},
            "output": {"answer": "A company that creates larg language models"},
        },
        {
            "input": {"question": "What is Google?"},
            "output": {"answer": "A technology company known for search"},
        },
    ]
)
```

```python
# Optional: Verify the dataset has the examples
refreshed_dataset = client.read_dataset(dataset_id=dataset.id)
print(f"Dataset '{refreshed_dataset.name}' now has examples. You can check the UI.")
```

# Define Metrics (LLM as a judge)

```python
from google import genai
API_KEY = os.environ['GEMINI_API_KEY']
GEMINI_MODEL = os.environ['GEMINI_MODEL']
from langsmith import wrappers

client = wrappers.wrap_gemini(genai.Client(api_key=API_KEY))

    # Basic text generation:
response = client.models.generate_content(
    model=GEMINI_MODEL,
    contents="What is langchain?",
)
print(response.text)
```

```python

client = genai.Client(api_key=os.environ['GEMINI_API_KEY'])

interaction = client.interactions.create(
    model=GEMINI_MODEL,
    input="Explain how AI works in a few words"
)
print(interaction.output_text)
```
